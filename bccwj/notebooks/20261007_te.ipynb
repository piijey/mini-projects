## 元記事を形態素解析する

MeCab + UniDic で、各文の最後の4語を見る (`data/hatelabo_20261005100640.txt`)

```python
import MeCab
import unidic

mecab = MeCab.Tagger(f"-d {unidic.DICDIR}")


def tokenize(text):
    """(書字形, 品詞) のリストにする。品詞は大分類-中分類"""
    tokens = []
    for row in mecab.parse(text).splitlines():
        if row == "EOS":
            break
        surface, feature = row.split("\t", 1)
        pos = feature.split(",")
        tokens.append((surface, f"{pos[0]}-{pos[1]}"))
    return tokens


with open("../data/hatelabo_20261005100640.txt") as f:
    sentences = [line.strip() for line in f if line.strip()]

for sentence in sentences:
    tail = tokenize(sentence)[-4:]
    print(" | ".join(f"{surface} ({pos})" for surface, pos in tail))
```

✏️メモ

| # | 文末 | 句点の2つ前の語 | 句点の直前の語 |
| -- | -- | -- | -- |
| 1 | 気がしていて。 | い（動詞-非自立可能） | て（助詞-接続助詞）👈 |
| 2 | 多すぎると思い。 | と（助詞-接続助詞） | 思い（名詞-普通名詞） |
| 3 | 疑問に思っていて。 | い（動詞-非自立可能） | て（助詞-接続助詞）👈 |
| 4 | 探したくて。 | たく（助動詞） | て（助詞-接続助詞）👈 |
| 5 | とすら思い。 | すら（助詞-副助詞） | 思い（名詞-普通名詞） |
| 6 | 本当に疑問で。 | 疑問（名詞-普通名詞） | で（助詞-格助詞） |
| 7 | 思ってしまって。 | しまっ（動詞-非自立可能） | て（助詞-接続助詞）👈 |
| 8 | 探していて。 | い（動詞-非自立可能） | て（助詞-接続助詞）👈 |
| 9 | 聞きたいと思い。 | と（助詞-格助詞） | 思い（名詞-普通名詞） |
| 10 | 疑問で。 | 疑問（名詞-普通名詞） | で（助詞-格助詞） |

- 接続助詞「て」で終わるのは 10 文中 5 文（今回 BCCWJ の検索（接続助詞 + 句点）で対象とするもの）
    - 「て」の直前は「い」（〜ている）3、「たく」（〜たい）1、「しまっ」（〜てしまう）1
- ほかは接続助詞ではない
    - 「思い。」(3) は名詞（と解析されているが実際には動詞）
    - 「疑問で。」 (2) の「で」は格助詞
